# 02. Pembagian Dataset Final (Train-Test Split)

Notebook ini membuat pembagian **train-test final** pada dataset **Diabetes Health Indicators (BRFSS 2015)**.

Isi dan logika pembagian mengikuti script sumber `create_final_split.py`.

## 0. Mengatur Folder Kerja

Notebook ini disimpan di folder `notebooks`, sedangkan data dan artifacts berada di folder `glysense_backend`. Sel berikut memindahkan folder kerja satu tingkat ke atas agar jalur seperti `data/...` dan `artifacts/...` ditemukan. Aman dijalankan berulang.

```python
import os

if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")

print(os.getcwd())
```

## 1. Membaca Dataset Asli

Jalur file mengasumsikan notebook dijalankan dari direktori utama `glysense_backend`.

```python
import os
import pandas as pd

from sklearn.model_selection import train_test_split

file_path = "data/diabetes_012_health_indicators_BRFSS2015.csv"

df = pd.read_csv(file_path)
```

## 2. Memisahkan Fitur dan Target

```python
X = df.drop(columns=["Diabetes_012"])

y = df["Diabetes_012"].astype(int)
```

## 3. Membuat Train-Test Split Final

Random state ini digunakan sebagai pembagian **FINAL** penelitian. Test set setelah ini tidak digunakan sampai evaluasi akhir.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=2026,
    stratify=y
)
```

## 4. Menggabungkan Fitur dan Target Kembali

```python
train_final = X_train.copy()
train_final["Diabetes_012"] = y_train

test_final = X_test.copy()
test_final["Diabetes_012"] = y_test
```

## 5. Membuat Folder Data Final

```python
output_folder = "data/final_split"

os.makedirs(
    output_folder,
    exist_ok=True
)
```

## 6. Menyimpan Data Train dan Test Final

```python
train_path = os.path.join(
    output_folder,
    "train_final.csv"
)

test_path = os.path.join(
    output_folder,
    "test_final_locked.csv"
)

train_final.to_csv(
    train_path,
    index=False
)

test_final.to_csv(
    test_path,
    index=False
)
```

## 7. Menampilkan Informasi Train Final

```python
print("=" * 70)
print("FINAL TRAIN-TEST SPLIT BERHASIL DIBUAT")
print("=" * 70)

print("\nJumlah seluruh data:")
print(len(df))

print("\nJumlah data train final:")
print(len(train_final))

print("\nJumlah data test final:")
print(len(test_final))

print("\nDistribusi kelas TRAIN FINAL:")
print(
    train_final["Diabetes_012"]
    .value_counts()
    .sort_index()
)

print("\nPersentase kelas TRAIN FINAL:")
print(
    train_final["Diabetes_012"]
    .value_counts(normalize=True)
    .sort_index()
    * 100
)

print("\nFile train disimpan di:")
print(train_path)

print("\nFile test final disimpan di:")
print(test_path)

print("\nPERINGATAN:")
print(
    "test_final_locked.csv jangan digunakan untuk "
    "training, tuning, cross-validation, threshold tuning, "
    "atau calibration."
)
```

## 8. Catatan

Notebook ini adalah konversi dari `create_final_split.py` tanpa perubahan logika. Pembagian bersifat deterministik (`random_state=2026`, `stratify=y`), sehingga hasilnya identik dengan pembagian final yang sudah ada.

File `test_final_locked.csv` tidak boleh digunakan untuk training, tuning, cross-validation, threshold tuning, maupun calibration. Test set hanya dipakai pada evaluasi akhir.
