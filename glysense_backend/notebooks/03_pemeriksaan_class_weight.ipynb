## 1. Membaca Dataset

Jalur file mengasumsikan notebook dijalankan dari direktori utama `glysense_backend`.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight

file_path = "data/diabetes_012_health_indicators_BRFSS2015.csv"

df = pd.read_csv(file_path)
```

## 2. Memisahkan Fitur dan Target

```python
X = df.drop(columns=["Diabetes_012"])
y = df["Diabetes_012"]
```

## 3. Train-Test Split

Catatan: pembagian ini menggunakan `random_state=42` sesuai script asli, sehingga **berbeda** dari pembagian final penelitian (`random_state=2026`). Pembagian ini hanya untuk pemeriksaan class weight dan tidak menyimpan file apa pun.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

## 4. Menghitung Class Weight

Dengan `class_weight="balanced"`, bobot tiap kelas berbanding terbalik dengan frekuensinya pada data training.

```python
classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = dict(zip(classes, weights))
```

## 5. Menampilkan Hasil

```python
print("Jumlah data training per kelas:")
print(y_train.value_counts().sort_index())

print("\nClass weight:")
for kelas, bobot in class_weights.items():
    print(f"Kelas {int(kelas)} = {bobot:.4f}")
```
